```python
import mlflow

import numpy as np
import pandas as pd
from sklearn import datasets, metrics
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import train_test_split

# import mlflow
# from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from scipy.stats import uniform
from sklearn.model_selection import RandomizedSearchCV
```

```python

# def eval_metrics(pred, actual):
#     rmse = np.sqrt(metrics.mean_squared_error(actual, pred))
#     mae = metrics.mean_absolute_error(actual, pred)
#     r2 = metrics.r2_score(actual, pred)
#     return rmse, mae, r2


def eval_metrics(y_true, y_pred):
    rmse = mean_squared_error(y_true, y_pred, squared=False)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    return rmse, mae, r2



# Set th experiment name
mlflow.set_experiment("wine-quality")

# Enable auto-logging to MLflow
mlflow.sklearn.autolog()

# Load wine quality dataset
X, y = datasets.load_wine(return_X_y=True)
print(pd.DataFrame(X).head())
print(y)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)

# # Start a run and train a model... default
# with mlflow.start_run(run_name="default-params"):
#     # mlflow.log_metric("metric_name", 123)
    
#     lr = ElasticNet()
#     lr.fit(X_train, y_train)

#     y_pred = lr.predict(X_test)
#     metrics = eval_metrics(y_pred, y_test)
```

```python

# Start a run and train a model... default
with mlflow.start_run(run_name="default-params"):
    # mlflow.log_metric("metric_name", 123)
    
    lr = ElasticNet()
    lr.fit(X_train, y_train)

    y_pred = lr.predict(X_test)
    metrics = eval_metrics(y_pred, y_test)
```

```python
# import mlflow
# from sklearn.linear_model import ElasticNet
# from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
# from scipy.stats import uniform
# from sklearn.model_selection import RandomizedSearchCV

# def eval_metrics(y_true, y_pred):
#     rmse = mean_squared_error(y_true, y_pred, squared=False)
#     mae = mean_absolute_error(y_true, y_pred)
#     r2 = r2_score(y_true, y_pred)
#     return rmse, mae, r2

lr = ElasticNet()

# Define distribution to pick parameter values from
distributions = dict(
    alpha=uniform(loc=0, scale=10),  # corrected to match actual uniform distribution [0, 10]
    l1_ratio=uniform()  # sample l1_ratio uniformly from [0, 1.0]
)

# Initialize random search instance
clf = RandomizedSearchCV(
    estimator=lr,
    param_distributions=distributions,
    scoring="neg_mean_absolute_error",
    cv=5,
    n_iter=100,
)

# Start a parent run in MLflow
with mlflow.start_run(run_name="hyperparameter-tuning"):
    search = clf.fit(X_train, y_train)

    # Evaluate the best model on test dataset
    y_pred = clf.best_estimator_.predict(X_test)
    rmse, mae, r2 = eval_metrics(y_test, y_pred)
    mlflow.log_metrics({
        "mean_squared_error_X_test": rmse,
        "mean_absolute_error_X_test": mae,
        "r2_score_X_test": r2,
    })
```

### use the endpoint to consume the service created

```python
import requests
import json

# Assuming 'df' is your DataFrame and 'label' is the name of your label column
# data = df.drop('label', axis=1).head(10)  # replace 'label' with the actual label column name

data = pd.DataFrame(X).head()

# Format the data as a dictionary under the key 'dataframe_split'
payload = {
    "dataframe_split": {
        "columns": data.columns.tolist(),
        "index": data.index.tolist(),
        "data": data.values.tolist()
    }
}

# Convert the payload to a JSON string
json_payload = json.dumps(payload)

# The endpoint where the model is being served
url = 'http://127.0.0.1:5009/invocations'

# The headers should indicate that you're sending JSON
headers = {
    'Content-Type': 'application/json',
}

# Send the POST request with the JSON payload
response = requests.post(url, headers=headers, data=json_payload)

# Check the status code first
if response.status_code == 200:
    # Decode the JSON response
    predictions = response.json()
    print(predictions)
else:
    # If the response was not 200, output the error code and message
    print(f"Error code from server: {response.status_code}")
    print(response.text)
```
